> **ADVERTENCIA**
> **`np.random.seed()` no es suficiente**
>
> Fijar la semilla global de NumPy **no** garantiza reproducibilidad en
> scikit-learn. Cada función que use azar necesita su propio `random_state`:
> `train_test_split(..., random_state=42)`, `KFold(..., random_state=42)`,
> `RandomForestClassifier(random_state=42)`. Si omites uno, obtendrás números
> distintos en cada ejecución y no sabrás si una mejora es real o es ruido.

## Sobre el uso de inteligencia artificial

Puedes usar herramientas de IA en este curso. De hecho, deberías: forman parte
del trabajo real de un analista. Pero el material está diseñado para que copiar
un enunciado a un chatbot **no resuelva la tarea**: casi todas las preguntas
piden interpretar *tu* salida concreta, justificar *tu* decisión en *tu*
escenario, o encontrar el error en un fragmento específico.

Las reglas de uso, qué debes declarar y qué se considera fraude académico están
en [Uso responsable de IA](https://wilsonsr.github.io/tecnicas-modelos-supervisados/recursos/uso-de-ia.html).

Lo que debes recordar

- Este sitio **apoya** los encuentros y es material de consulta; las actividades calificadas viven en el aula virtual.

- Todos los cuadernillos siguen la misma ruta: problema → datos → exploración → pregunta → modelo → evaluación → interpretación → decisión.

- Reproducibilidad = semilla fija en *cada* función aleatoria + rutas relativas + cuaderno ejecutable de principio a fin.

- Dos datasets recorren todo el curso: vivienda en Bogotá (regresión) y ausentismo laboral (clasificación).

- Las microactividades no se califican, pero sin ellas las actividades que sí se califican se vuelven difíciles.
