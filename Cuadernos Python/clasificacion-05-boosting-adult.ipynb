```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["clase_trabajo"] = le.fit_transform(df["clase_trabajo"])
df["educacion"] = le.fit_transform(df["educacion"])
df["estado_civil"] = le.fit_transform(df["estado_civil"])
df["ocupacion"] = le.fit_transform(df["ocupacion"])
df["Parentesco"] = le.fit_transform(df["Parentesco"])
df["Genero"] = le.fit_transform(df["Genero"])
df["Raza"] = le.fit_transform(df["Raza"])
df["pais_origen"] = le.fit_transform(df["pais_origen"])
df.head(5)
```

# ¿Qué es Light GBM?

Light GBM es un marco de refuerzo de gradiente rápido, distribuido y de alto rendimiento basado en un algoritmo de árbol de decisión, que se utiliza para  clasificar y muchas otras tareas de aprendizaje automático.

Dado que se basa en algoritmos de árboles de decisión, divide la hoja del árbol con el mejor ajuste, mientras que otros algoritmos de impulso dividen el árbol en profundidad o en nivel en lugar de en hoja. Por lo tanto, cuando crece en la misma hoja en Light GBM, el algoritmo de hoja puede reducir más pérdidas que el algoritmo de nivel y, por lo tanto, da como resultado una precisión mucho mejor que rara vez se puede lograr con cualquiera de los algoritmos de refuerzo existentes. Además, es sorprendentemente muy rápido, de ahí la palabra 'Light'.


- Xgboost

- Crecimiento de hojas con Light GBM

Las divisiones por hojas conducen a un aumento de la complejidad y pueden conducir a un sobreajuste y se pueden superar especificando otro parámetro profundidad máxima que especifica la profundidad a la que se producirá la división.



## Ventajas de Light GBM

- Velocidad de entrenamiento más rápida y mayor eficiencia : Light GBM utiliza un algoritmo basado en histogramas, es decir, agrupa valores de características continuos en contenedores discretos que aceleran el procedimiento de entrenamiento.
- Menor uso de memoria: reemplaza los valores continuos por contenedores discretos, lo que da como resultado un menor uso de memoria.

- Mejor precisión que cualquier otro algoritmo de refuerzo: produce árboles mucho más complejos siguiendo un enfoque de división por hojas en lugar de un enfoque por niveles, que es el factor principal para lograr una mayor precisión. Sin embargo, a veces puede dar lugar a un sobreajuste que se puede evitar configurando el parámetro max_depth.

- Compatibilidad con grandes conjuntos de datos: es capaz de funcionar igual de bien con grandes conjuntos de datos con una reducción significativa en el tiempo de entrenamiento en comparación con XGBOOST.


## Parámetros de ajuste de Light GBM
Light GBM utiliza la división por hojas en lugar de la división por profundidad, lo que le permite converger mucho más rápido, pero también conduce a un sobreajuste.

### Para un mejor ajuste

- num_leaves : Este parámetro se utiliza para establecer el número de hojas que se formarán en un árbol. Teóricamente, la relación entre num_hojas y max_ depth es num_leaves= 2^(max_ depth). Sin embargo, esta no es una buena estimación en el caso de GBM liGHT, ya que la división se realiza en forma de hoja y no en profundidad. Por lo tanto, el conjunto num_leaves debe ser menor que 2^(max_ depth), de lo contrario, puede provocar un ajuste excesivo. Light GBM no tiene una relación directa entre num_leaves y max_ depth y, por lo tanto, los dos no deben estar vinculados entre sí.
- min_data_in_leaf : también es uno de los parámetros importantes para tratar el sobreajuste. Establecer su valor más pequeño puede causar un sobreajuste y, por lo tanto, debe establecerse en consecuencia. Su valor debe ser de cientos a miles de grandes conjuntos de datos.
- max_ depth : Especifica la profundidad máxima o el nivel hasta el que puede crecer el árbol.


## Para mayor velocidad
- bagging_fraction : se utiliza para realizar embolsado para obtener resultados más rápidos
- feature_fraction : establece la fracción de las funciones que se usarán en cada iteración
- max_bin : un valor más pequeño de max_bin puede ahorrar mucho tiempo, ya que clasifica los valores de las características en contenedores discretos, lo que es computacionalmente económico.


## Para una mejor precisión
Usa datos de entrenamiento más grandes
- num_leaves : configurarlo en un valor alto produce árboles más profundos con mayor precisión, pero conduce a un sobreajuste. Por lo tanto, no se prefiere su valor más alto.
- max_bin : configurarlo en valores altos tiene un efecto similar al causado por el aumento del valor de num_leaves y también más lento nuestro procedimiento de entrenamiento.
